```python
# ============================================================
# CUSTOMER REVIEW CLEANING SYSTEM - NLP
# ============================================================

import pandas as pd
import re
import string
import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer, PorterStemmer

# ============================================================
# 1. DOWNLOAD NLP RESOURCES
# ============================================================

nltk.download("stopwords", quiet=True)
nltk.download("wordnet", quiet=True)

# ============================================================
# 2. LOAD DATASET
# ============================================================

file_path = "NLPTASKS/12thaugCustomer_Sentiment.csv"

df = pd.read_csv(file_path)

review_column = "review_text"

print("=" * 70)
print("CUSTOMER REVIEW CLEANING SYSTEM")
print("=" * 70)

print("\nOriginal Dataset Shape:", df.shape)
print("Review Column:", review_column)

# ============================================================
# 3. DATA CLEANING
# ============================================================

# Remove missing review values
df = df.dropna(subset=[review_column]).copy()

print("After Removing Missing Reviews:", df.shape)

# ============================================================
# 4. INITIALIZE NLP TOOLS
# ============================================================

stop_words = set(stopwords.words("english"))

lemmatizer = WordNetLemmatizer()

stemmer = PorterStemmer()

# ============================================================
# 5. LOWERCASE
# ============================================================

df["lowercase"] = (
    df[review_column]
    .astype(str)
    .str.lower()
)

# ============================================================
# 6. REMOVE HTML TAGS
# ============================================================

df["without_html"] = df["lowercase"].apply(
    lambda x: re.sub(r"<[^>]*>", " ", x)
)

# ============================================================
# 7. REMOVE URLs
# ============================================================

df["without_urls"] = df["without_html"].apply(
    lambda x: re.sub(
        r"https?://\S+|www\.\S+",
        " ",
        x
    )
)

# ============================================================
# 8. REMOVE EMAIL ADDRESSES
# ============================================================

df["without_emails"] = df["without_urls"].apply(
    lambda x: re.sub(
        r"\b[\w.-]+@[\w.-]+\.\w+\b",
        " ",
        x
    )
)

# ============================================================
# 9. REMOVE PUNCTUATION
# ============================================================

df["without_punctuation"] = df["without_emails"].apply(
    lambda x: x.translate(
        str.maketrans("", "", string.punctuation)
    )
)

# ============================================================
# 10. REMOVE NUMBERS
# ============================================================

df["without_numbers"] = df["without_punctuation"].apply(
    lambda x: re.sub(r"\d+", " ", x)
)

# ============================================================
# 11. REMOVE SPECIAL CHARACTERS
# ============================================================

df["without_special_characters"] = df["without_numbers"].apply(
    lambda x: re.sub(
        r"[^a-zA-Z\s]",
        " ",
        x
    )
)

# ============================================================
# 12. REMOVE EXTRA SPACES
# ============================================================

df["normalized_text"] = df["without_special_characters"].apply(
    lambda x: re.sub(
        r"\s+",
        " ",
        x
    ).strip()
)

# ============================================================
# 13. TOKENIZATION
# ============================================================

df["tokens"] = df["normalized_text"].apply(
    lambda x: x.split()
)

# ============================================================
# 14. STOPWORD REMOVAL
# ============================================================

df["without_stopwords"] = df["tokens"].apply(
    lambda words: [
        word
        for word in words
        if word not in stop_words
    ]
)

# ============================================================
# 15. STEMMING
# ============================================================

df["stemmed_tokens"] = df["without_stopwords"].apply(
    lambda words: [
        stemmer.stem(word)
        for word in words
    ]
)

# ============================================================
# 16. LEMMATIZATION
# ============================================================

df["lemmatized_tokens"] = df["without_stopwords"].apply(
    lambda words: [
        lemmatizer.lemmatize(word)
        for word in words
    ]
)

# ============================================================
# 17. FINAL CLEANED REVIEW
# ============================================================

df["cleaned_review"] = df["lemmatized_tokens"].apply(
    lambda words: " ".join(words)
)

# ============================================================
# 18. DISPLAY TWO EXAMPLE REVIEWS
# ============================================================

print("\n")
print("=" * 70)
print("EXAMPLE REVIEWS - NLP PROCESSING")
print("=" * 70)

# Take two different reviews for demonstration
examples = df.head(2)

for index, row in examples.iterrows():

    print("\n" + "=" * 70)
    print("EXAMPLE REVIEW")
    print("=" * 70)

    print("\n1. ORIGINAL REVIEW")
    print(row[review_column])

    print("\n2. LOWERCASE")
    print(row["lowercase"])

    print("\n3. HTML REMOVAL")
    print(row["without_html"])

    print("\n4. URL REMOVAL")
    print(row["without_urls"])

    print("\n5. EMAIL REMOVAL")
    print(row["without_emails"])

    print("\n6. PUNCTUATION REMOVAL")
    print(row["without_punctuation"])

    print("\n7. NUMBER REMOVAL")
    print(row["without_numbers"])

    print("\n8. SPECIAL CHARACTER REMOVAL")
    print(row["without_special_characters"])

    print("\n9. NORMALIZATION / EXTRA SPACE REMOVAL")
    print(row["normalized_text"])

    print("\n10. TOKENIZATION")
    print(row["tokens"])

    print("\n11. STOPWORD REMOVAL")
    print(row["without_stopwords"])

    print("\n12. STEMMING")
    print(row["stemmed_tokens"])

    print("\n13. LEMMATIZATION")
    print(row["lemmatized_tokens"])

    print("\n14. FINAL CLEANED REVIEW")
    print(row["cleaned_review"])


# ============================================================
# 19. SAVE PROCESSED DATASET
# ============================================================

output_file = "NLPTASKS/customer_reviews_cleaned.csv"

df.to_csv(
    output_file,
    index=False
)

# ============================================================
# 20. FINAL INFORMATION
# ============================================================

print("\n")
print("=" * 70)
print("PROCESSING COMPLETED SUCCESSFULLY")
print("=" * 70)

print("Final Dataset Shape:", df.shape)
print("Saved File:", output_file)

print("\nProcessing Techniques Applied:")
print("✓ Missing Value Removal")
print("✓ Lowercasing")
print("✓ HTML Removal")
print("✓ URL Removal")
print("✓ Email Removal")
print("✓ Punctuation Removal")
print("✓ Number Removal")
print("✓ Special Character Removal")
print("✓ Extra Space Removal")
print("✓ Tokenization")
print("✓ Stopword Removal")
print("✓ Stemming")
print("✓ Lemmatization")
print("✓ Final Cleaned Review")

print("=" * 70)
```
