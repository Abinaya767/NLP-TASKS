```python
import re
from collections import Counter, defaultdict

path = "next_word_predictor.txt"

with open(path, "r", encoding="utf-8") as f:
    text = f.read().lower()

text = re.sub(r"\b\w+'(s|re|ve|ll|d|m|t)\b", "", text)
text = re.sub(r"[^a-z\s]", " ", text)
text = re.sub(r"\s+", " ", text)

words = text.split()

predictor = defaultdict(Counter)

for i in range(len(words) - 1):
    predictor[words[i]][words[i + 1]] += 1

word = input("Enter a word: ").lower().strip()
print(word)
print("\nNext possible words:")

if word in predictor:
    for w, count in predictor[word].most_common(5):
        print(w)
else:
    print("Word not found")
```
