```python
import pandas as pd
import re
import nltk
from nltk.tokenize import word_tokenize
from nltk.metrics import edit_distance
nltk.download("punkt")
nltk.download("punkt_tab")
df = pd.read_csv("spelling_dataset.csv")
print("First 5 records:")
print(df.head())

# Create vocabulary from correct words
vocabulary = set()
for word in df["correct_word"]:
    vocabulary.add(str(word).lower())
print("\nVocabulary size:", len(vocabulary))

query = input("\nEnter your search query: ")
print("\nOriginal Query:")
print(query)
query = query.lower()

# Tokenize
tokens = word_tokenize(query)
# Keep only words
tokens = [word for word in tokens if re.match(r"^[a-zA-Z]+$", word)]
print("\nTokens:")
print(tokens)

# Find incorrect words
incorrect_words = []
for word in tokens:
    if word not in vocabulary:
        incorrect_words.append(word)
print("\nIncorrect words:")
print(incorrect_words)

# Find corrections
corrections = {}
for incorrect_word in incorrect_words:
    best_word = None
    minimum_distance = float("inf")
    for candidate in vocabulary:

        distance = edit_distance(incorrect_word, candidate)

        if distance < minimum_distance:
            minimum_distance = distance
            best_word = candidate

    corrections[incorrect_word] = best_word

# Display suggestions
print("\nSuggested Corrections:")
for incorrect, corrected in corrections.items():
    print(incorrect, "->", corrected)
    
# Create corrected query
corrected_tokens = []
for word in tokens:
    if word in corrections:
        corrected_tokens.append(corrections[word])
    else:
        corrected_tokens.append(word)
corrected_query = " ".join(corrected_tokens)

# Final output
print("\n-----------------------------")
print("Original Query:")
print(query)

print("\nIncorrect Words:")
print(incorrect_words)

print("\nCorrections:")
print(corrections)

print("\nFinal Corrected Query:")
print(corrected_query)
print("-----------------------------")
```
